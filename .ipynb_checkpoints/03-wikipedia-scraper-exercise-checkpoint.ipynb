### open the web site with requests

```python
url = "https://en.wikipedia.org/wiki/Category:Sufi_mysti"
# Your identification
headers = {"user-agent" : "Mozilla/5.0 (Windows NT 6.1) AppleWebKit/537.36(KHTML, like Gecko) Chrome/41.0.2228.0 Safari/537.36;",
"from": "Ashley Mowreader amowreader@protonmail.com"}
page = requests.get(url, headers=headers)
page_content = page.content
```

### parse the page through the BeautifulSoup library

```python
soup = BeautifulSoup(page_content, "html.parser")

## list = soup.find("html element" {"html element":"specific name"})
all_groupings = soup.find("div", {"class":"mw-category"})
print(all_groupings.prettify())
```
