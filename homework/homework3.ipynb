```python
import csv
from bs4 import BeautifulSoup
```

```python
# Open and read the html file (remember that your file path may be different!!!)
with open("data/gutenberg.html", "r") as page:
     # Tell the BeautifulSoup library to interpret the file
    soup = BeautifulSoup(page, "html.parser")
```

```python
print(soup)
```

```python
  # Find the first h1 on the page. Print the text stored within it.
first_h1 = soup.find("h1")

## to get text use .get_text or .text (text is depriciated) 
print(first_h1.prettify())
```

```python
    # There are three follow links on the page, one for Facebook, one for Bluesky, and one for Mastodon. 
    # What is the class associated with these follow links? Answer here:
    ## <ul class="pgx-social">

    # What html element are these follow links stored in? Answer here: 
   ## <a class="is-facebook">
    ## <a class="is-mastodon">
    ## <a class="is-bluesky">

social_links=[]
    # Use the function soup.find_all(...) to grab the html associated with the three follow links and store them in an array. 
social_links = soup.find_all("a", class_=["is-facebook", "is-mastodon", "is-bluesky"])
# Loop through the list, printing the text for each follow link.
for link in social_links:
    print(link["href"])
```
