```python
##importing pandas

import pandas as pd
```

```python
##building my df 
data = pd.read_csv("data/copper-duty_annual.csv")
```

```python
##is this as ugly as i think it is ?
data.shape
```

```python
##answer yes, lets clean this mofo 
data=data.reset_index()
```

```python
## transposing the data 
np_array = data.to_numpy().T
copper = pd.DataFrame(np_array, index=data.columns, columns=data.index)
```

```python
## resetting the column heads and cleaning the index 
copper.columns = copper.iloc[0]

# resetting the index and cleaning 
copper = copper.drop(copper.index[0])
copper = copper.reset_index(drop=True)
```

```python
copper.tail(10)
```

```python
copper[['HTS Code', 'Product Name']] = copper['Commodity'].str.split(' ',n=1, expand=True)
```

```python
copper.head(10)
```

```python
copper.info()
```

```python
copper.to_csv("export-copper-duty-annual.csv", index=False)
```

```python
## renaming the column bc its ugly 

copper["Calculated Duty"] = copper["Calculated Duty($US)"]
```

```python
## making Calculated Duty a float instead of a string 
copper["Calculated Duty"]= copper["Calculated Duty"].astype(float)
```

```python
## ok what's cooking good looking

print(copper["Calculated Duty"].mean())
print(copper["Calculated Duty"].median())
print(copper["Calculated Duty"].max())
```

```python
## well i want to learn the change over time so lets index this 
indexed_copper = copper.set_index("Time")
indexed_copper
```
