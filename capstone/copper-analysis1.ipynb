```python
##importing pandas
import pandas as pd
```

```python
##building my df 
data = pd.read_csv("data/copper-duty_annual.csv")
```

```python
##is this as ugly as i think it is ?
data.shape
```

```python
##answer yes, lets clean this mofo 
data=data.reset_index()
```

```python
## transposing the data 
np_array = data.to_numpy().T
copper = pd.DataFrame(np_array, index=data.columns, columns=data.index)
```

```python
## resetting the column heads and cleaning the index 
copper.columns = copper.iloc[0]

# resetting the index and cleaning 
copper = copper.drop(copper.index[0])
copper = copper.reset_index(drop=True)
```

```python
copper.tail(10)
```

```python
copper[['HTS Code', 'Product Name']] = copper['Commodity'].str.split(' ',n=1, expand=True)
```

```python
copper.head(10)
```

```python
copper.info()
```

```python
copper.to_csv("export-copper-duty-annual.csv", index=False)
```

```python
## renaming the column bc its ugly 

copper["Calculated Duty"] = copper["Calculated Duty($US)"]
```

```python
## making Calculated Duty a float instead of a string 
copper["Calculated Duty"]= copper["Calculated Duty"].astype(float)
```

```python
## ok what's cooking good looking

print(copper["Calculated Duty"].mean())
print(copper["Calculated Duty"].median())
print(copper["Calculated Duty"].max())
```

```python
## well i want to learn the change over time so lets index this 
indexed_copper = copper.set_index("Time")
indexed_copper
```

```python
## lets index by calculated duty to see  

copper.groupby(
    ["HTS Code","Product Name","Time"]
)["Calculated Duty"].median().reset_index().sort_values(by="Calculated Duty",ascending=False)
```

```python
## ok so I can say right now that the calculated duty on all copper pipes comming into the U.S. grew like, a lot. From $159M in 2025 to $220M in early 2026.
## lets look at the monthly dataaaa

# faster load for this very wide file
monthly_data = pd.read_csv(
    "data/copper-duty-values-by_month.csv",
    skiprows=1,
    header=None,
    dtype=str
 )

monthly_data = monthly_data.set_index(0).T.reset_index(drop=True)
monthly_data.columns.name = None

monthly_data["Calculated Duty($US)"] = pd.to_numeric(
    monthly_data["Calculated Duty($US)"], errors="coerce"
 )

monthly_data[["Month", "Year"]] = monthly_data["Time"].str.split(" ", n=1, expand=True)

monthly_data.head(
)
```

```python
## calculating the first 8 months of 2025 imports
ytd_2025 = monthly_data[(monthly_data["Year"] == '2025') & (monthly_data["Month"].isin([
        "January", "February", "March", "April",
        "May", "June", "July", "August"
    ]))]["Calculated Duty($US)"].sum()
print(f"YTD 2025 (Jan-Aug): ${ytd_2025:,.0f}")

## calculating the first 8 months of 2026 imports
total_2026 = monthly_data[(monthly_data["Year"] == '2026') & (monthly_data["Month"].isin([
        "January", "February", "March", "April",
        "May", "June", "July", "August"
    ]))]["Calculated Duty($US)"].sum()
print(f"Total 2026 (Jan-Aug): ${total_2026:,.0f}")
```
