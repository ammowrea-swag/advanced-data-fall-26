```python
##building my df 
data = pd.read_csv("data/copper-customs-value_annualized.csv")
```

```python
##answer yes, lets clean this mofo 
data=data.reset_index()
```

```python
## transposing the data 
np_array = data.to_numpy().T
data = pd.DataFrame(np_array, index=data.columns, columns=data.index)
```

```python
## resetting the column heads and cleaning the index 
data.columns = data.iloc[0]
```

```python
data
```

```python

# resetting the index and cleaning 
data = data.drop(data.index[0])

data = data.reset_index(drop=True)
```

```python
## ok lets see if it looks pretty now ! 
data.head(10)
```

```python
## breaking out the HTS code from the Commodity information so I can do analysis 
data[['HTS Code', 'Product Name']] = data['Commodity'].str.split(' ',n=1, expand=True)
```

```python
##copper.to_csv("export-copper-customs_value-monthly.csv", index=False)

```

```python
## renaming the column bc its ugly 

data["CIF Value"] = data["CIF Value (Gen) ($US)"]
```

```python
data["CIF Value"]= data["CIF Value"].astype(float)
```

```python
## ok lets see what she looks like inside
print(data["CIF Value"].mean())
print(data["CIF Value"].median())
print(data["CIF Value"].max())
```

```python
## summing all tariffs on copper for 2026
total_2026 = data.loc[data["Time"] == '2026 through August', "CIF Value"].sum()
print(f"2026: ${total_2026:,.0f}")
```

```python
## summing all tariffs on copper for 2025
total_2025 = data.loc[data["Time"] == '2025', "CIF Value"].sum()
print(f"2025: ${total_2025:,.0f}")
```

```python
##ugh I want to do a comparison of imports by month, so I need to build a new df for that
monthly_data = pd.read_csv("data/copper-customs-value_monthly.csv")
monthly_data.shape
```

```python
monthly_data = pd.read_csv("data/copper-customs-value_monthly.csv", skiprows=1, header=None)

monthly_data = monthly_data.set_index(0).T.reset_index(drop=True)
monthly_data.columns.name = None

monthly_data["CIF Value (Gen) ($US)"] = pd.to_numeric(
    monthly_data["CIF Value (Gen) ($US)"], errors="coerce"
)

monthly_data.head()
```

```python
monthly_data[["Month", "Year"]] = monthly_data["Time"].str.split(" ", n=1, expand=True)

monthly_data.head(
)
```

```python
## calculating the first 8 months of 2025 imports
ytd_2025 = monthly_data[(monthly_data["Year"] == '2025') & (monthly_data["Month"].isin([
        "January", "February", "March", "April",
        "May", "June", "July", "August"
    ]))]["CIF Value (Gen) ($US)"].sum()
print(ytd_2025)
print(f"YTD 2025 (Jan-Aug): ${ytd_2025:,.0f}")
```
