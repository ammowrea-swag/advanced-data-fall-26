```python
## building my df 
data = pd.read_csv(
    "data/country-duty.csv",
    skiprows=1,
    header=None,
    dtype=str
 )

## transposing the data into a clean format
data = data.set_index(0).T.reset_index(drop=True)
data.columns.name = None

data["Calculated Duty($US)"] = pd.to_numeric(
    data["Calculated Duty($US)"], errors="coerce" )

data.head()
```

```python
## how many countries I got in this baddie
data["Country"].nunique()
```

```python
## more cleaning, lets tidy up my columns so I can see the product code by itself 
data[['HTS Code', 'Product Name']] = data['Commodity'].str.split(' ',n=1, expand=True)
```

```python
data.head(10)
```

```python
## renaming the column bc its ugly 

data["Calculated Duty"] = data["Calculated Duty($US)"]
```

```python
## making Calculated Duty a float instead of a string 
data["Calculated Duty"]= data["Calculated Duty"].astype(float)
```

```python
## looking for the top importers of copper in 2025
top_country_2025 = (
    data.loc[pd.to_numeric(data["Time"], errors="coerce").eq(2025)]
    .groupby("Country", as_index=False)["Calculated Duty($US)"]
    .sum()
    .sort_values("Calculated Duty($US)", ascending=False)
    .set_index("Country")
)

top_country_2026 = (
    data.loc[pd.to_numeric(data["Time"], errors="coerce").eq(2026)]
    .groupby("Country", as_index=False)["Calculated Duty($US)"]
    .sum()
    .sort_values("Calculated Duty($US)", ascending=False)
    .set_index("Country")
)

```
