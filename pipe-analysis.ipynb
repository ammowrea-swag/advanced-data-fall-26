```python
## importingggg 
import pandas as pd
```

```python
## importing ! 
data = pd.read_csv("capstone/data/just-pipes.csv")
```

```python
##is this as ugly as i think it is ?
data.shape
```

```python
##answer yes, lets clean this mofo 
data=data.reset_index()
```

```python
## transposing the data 
np_array = data.to_numpy().T
pipes = pd.DataFrame(np_array, index=data.columns, columns=data.index)
```

```python
## resetting the column heads and cleaning the index 
pipes.columns = pipes.iloc[0]

# resetting the index and cleaning 
pipes = pipes.drop(pipes.index[0])
pipes = pipes.reset_index(drop=True)
```

```python
pipes.tail(10)
```

```python
pipes[['HTS Code', 'Product Name']] = pipes['Commodity'].str.split(' ',n=1, expand=True)
```

```python
pipes.head(10)
```

```python
pipes.info()
```

```python
pipes["Calculated Duty"] = pipes["Calculated Duty($US)"].astype(float)
```

```python
print(pipes["Calculated Duty"].mean())
print(pipes["Calculated Duty"].median())
print(pipes["Calculated Duty"].max())
```

```python
## well i want to learn the change over time so lets index this 
indexed_pipes = pipes.set_index("Time")
indexed_pipes
```

```python
## ok lets find out which continent has the most duties for pipes! 
pipes.groupby(
    ["HTS Code","Product Name","Country"]
)["Calculated Duty"].median().reset_index().sort_values(by="Calculated Duty",ascending=False)


#its Asia and Europe! So I'm going to download more data with just countries
```
