```python
data = pd.read_csv('../data/clean-payroll-data.csv',low_memory=False)

##importing the data
```

```python
##cleaning the string
data["Base Salary"]= data["Base Salary"].str.replace("$","").str.replace(",","").str.replace(".00","")
```

```python
##converting the base salary string to float 
data["base_salary_float"] = data["Base Salary"].astype(float)
```

```python
##looking for the median of each job descriptions job salary

data.groupby(
    ["Title Description"]
)["base_salary_float"].median().reset_index().sort_values(by="base_salary_float",ascending=False)
```

```python
##grouping by two columns 
data.groupby(
    ["Agency Name", "Work Location Borough"]
)["base_salary_float"].median().reset_index().sort_values(by="base_salary_float",ascending=False)
```

```python
data.groupby(
    ["Agency Name", "Work Location Borough"]).agg(
    {"base_salary_float":"median",
     "base_salary_float":"max",
     "base_salary_float":"min"
    }
).reset_index(
).sort_values(
    by="base_salary_float",
    ascending=False)
```
